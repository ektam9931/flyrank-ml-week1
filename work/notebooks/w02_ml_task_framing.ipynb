## 5. Why ML beats a fixed rule here

A fixed rule would be too simple because content performance depends on several signals at the same time, including search demand, impressions, clicks, engagement, ranking position, content age, and recent trends. Different pages can have different combinations of these signals, so one threshold may not work well for all pages. ML can learn patterns across these measured features and provide decision-support for which content may need attention.

```python
features = [
    "search_volume",
    "impressions_90d",
    "clicks_90d",
    "engagement_rate",
    "avg_position",
    "content_age_days",
    "trend_pct"
]

print("Number of performance signals:", len(features))
print("Signals used to support the ML framing:", features)
```
